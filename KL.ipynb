## STEP 5: OUTLIER DETECTION & TREATMENT

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Select numeric columns for outlier analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns
print("📊 Numeric columns:", list(numeric_cols))
```

```python
# 1️⃣ Detect outliers using the IQR method
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    before = data.shape[0]
    data = data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]
    after = data.shape[0]
    removed = before - after
    print(f"🧹 {column}: Removed {removed} outliers (bounds: {lower_bound:.2f} – {upper_bound:.2f})")
    return data
```

```python
# 2️⃣ Apply to important numeric columns (adjust as needed)
for col in ["size_sqft", "price_myr"]:
    if col in df.columns:
        df = remove_outliers_iqr(df, col)
```

```python
import numpy as np
from sklearn.base import BaseEstimator, RegressorMixin
from statsmodels.nonparametric.kernel_regression import KernelReg

class MixedKDERegressor(BaseEstimator, RegressorMixin):
    def __init__(self, var_type='c', reg_type='ll', bw='normal_reference'):
        self.var_type = var_type
        self.reg_type = reg_type
        self.bw = bw
        self.model_ = None
        self.bw_ = None  # <-- NEW: Attribute to store the final learned bandwidths!

    def fit(self, X, y):
        X_arr = np.asarray(X, dtype=float)
        y_arr = np.asarray(y, dtype=float)
        
        # Instantiate and fit the statsmodels KernelReg
        self.model_ = KernelReg(
            endog=y_arr, 
            exog=X_arr, 
            var_type=self.var_type, 
            reg_type=self.reg_type, 
            bw=self.bw
        )
        
        # ⚡ STORE THE CALCULATED BANDWIDTHS: 
        # Whether calculated via normal_reference, AIC, or manual arrays, 
        # statsmodels saves the final numbers in self.model_.bw
        self.bw_ = self.model_.bw
        
        return self

    def predict(self, X):
        X_arr = np.asarray(X, dtype=float)
        predictions, _ = self.model_.fit(X_arr)
        return predictions
```

```python
# --- 2️⃣ Define Models + Hyperparameter Grids ---
MY_VAR_TYPE = 'uccccuuu'  # <-- THIS LINE PREVENTS THE NAME ERROR!

models = {
    "Linear Regression": (LinearRegression(), {}),  # No hyperparameters to tune
    "Lasso Regression": (Lasso(random_state=42, max_iter=3000), {
        "alpha": [0.001, 0.01, 0.1, 1.0, 10.0, 100.0],  # Hyperparameter grid for alpha
    }),
    "Mixed-KDE Regression": (MixedKDERegressor(var_type=MY_VAR_TYPE), {
        # Tune between Local Linear ('ll') and Local Constant/Nadaraya-Watson ('lc')
        "reg_type": ["lc", "ll"],
        # Tune between Least-Squares Cross-Validation and AIC for bandwidth calculation
        "bw": ['normal_reference'],
    }),
    "Random Forest": (RandomForestRegressor(random_state=42, n_jobs=-1), {
        "n_estimators": [200, 300],
        "max_depth": [10, 20, None],
        "min_samples_split": [2, 5],
    }),
}
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import GridSearchCV

results = []
trained_models = {}  # Store fitted models

for name, (model, param_grid) in models.items():
    print(f"\n🚀 Training model: {name}")

    # 1. Dataset Routing
    if name in ["Linear Regression", "Lasso Regression"]:
        X_train, y_train = X_train_scaled, y_train_unscaled
        X_test, y_test = X_test_scaled, y_test_unscaled
    else:
        X_train, y_train = X_train_kde, y_train_unscaled
        X_test, y_test = X_test_kde, y_test_unscaled

    # 2. FITTING PHASE
    if param_grid:
        jobs = 1 if name == "Mixed-KDE Regression" else -1
        grid = GridSearchCV(model, param_grid, cv=3, n_jobs=jobs, scoring="neg_mean_squared_error")
        grid.fit(X_train, y_train)
        model = grid.best_estimator_
        print(f"🔧 Best Params: {grid.best_params_}")
    else:
        model.fit(X_train, y_train)

    trained_models[name] = model  # Save trained version

    # =====================================================================
    # ⚡ INSPECTION PHASE: Paste the Bandwidth Report right here!
    # It checks if the model is our KDE model, and if so, prints the weights.
    # =====================================================================
    if name == "Mixed-KDE Regression" and hasattr(model, "bw_"):
        print("\n📐 Learned Smoothing Bandwidths (bw) per Feature:")
        print("-" * 55)
        for col_name, bw_val, var_char in zip(X_train.columns, model.bw_, MY_VAR_TYPE):
            type_label = "Continuous ('c')" if var_char == 'c' else ("Unordered ('u')" if var_char == 'u' else "Ordered ('o')")
            print(f" -> [{type_label:<17}] {col_name:<18}: {bw_val:.4f}")
        print("-" * 55 + "\n")
    # =====================================================================

    # 3. TESTING / EVALUATION PHASE
    y_pred = model.predict(X_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    results.append({"Model": name, "RMSE": rmse, "MAE": mae, "R²": r2})
    print(f"{name} → RMSE: RM {rmse:,.2f}, MAE: RM {mae:,.2f}, R²: {r2:.3f}")
```
